```python
data = {'名前': ['アリス', 'ボブ', 'チャーリー'], '年齢': [25, 30, 35], '都市': ['東京', '大阪', '京都'], '年収(万)' : ['500', '300', '1500' ]}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
#棒グラフを描くためのツールを読み込み、pltと名付けた。
import matplotlib.pyplot as plt
#グラフをより簡単にきれいに書くための拡張ツールをsnsと名付け、読み込んだ。
import seaborn as sns

#フォント指定
plt.rcParams["font.family"] = "sans-serif"

#グラフの台紙の大きさ(インチ)
plt.figure(figsize=(8, 6))
#x横軸とy縦軸
sns.barplot(x="名前", y="年収(万)", data=df)
#グラフ名
plt.title("名前と年収の関係")
#以上の設定を画面下に表示
plt.show()
```

```python
data = {'名前': ['アリス', 'ボブ', 'チャーリー'], '年齢': [25, 30, 35], '都市': ['東京', '大阪', '京都'], '年収(万)' : ['500', '300', '1500' ]}
import pandas as pd

df = pd.DataFrame(data)

import matplotlib.pyplot as plt
import seaborn as sns

df["年収(万)"] = df["年収(万)"].astype(int)

plt.figure(figsize=(8, 6))
sns.barplot(x="名前", y="年収(万)", data=df)

plt.title("名前と年収の関係")
plt.show()
```
